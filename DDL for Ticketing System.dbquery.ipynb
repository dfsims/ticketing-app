```sql
CREATE SCHEMA IF NOT EXISTS ticketing_system.lakebase
COMMENT 'Schema for ticket management system';

-- Use the schema
USE SCHEMA ticketing_system.lakebase;

-- Create the tickets table
CREATE TABLE IF NOT EXISTS `tickets` (
  `ticket_id` BIGINT GENERATED ALWAYS AS IDENTITY,
  `title` STRING NOT NULL,
  `status` STRING NOT NULL,
  `created_by` STRING NOT NULL,
  `created_at` TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP(),
  PRIMARY KEY (`ticket_id`)
)
COMMENT 'Main tickets table for tracking support tickets';

-- Create the ticket_messages table with foreign key reference
CREATE TABLE IF NOT EXISTS `ticket_messages` (
  `message_id` BIGINT GENERATED ALWAYS AS IDENTITY,
  `ticket_id` BIGINT NOT NULL,
  `message_text` STRING NOT NULL,
  `author` STRING NOT NULL,
  `created_at` TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP(),
  PRIMARY KEY (`message_id`),
  CONSTRAINT `fk_ticket_messages_ticket_id` 
    FOREIGN KEY (`ticket_id`) 
    REFERENCES `tickets`(`ticket_id`)
)
COMMENT 'Messages associated with tickets';

-- Create an index on ticket_id for better query performance
CREATE INDEX IF NOT EXISTS `idx_ticket_messages_ticket_id` 
ON `ticket_messages` (`ticket_id`);
```
